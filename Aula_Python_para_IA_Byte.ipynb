## 🔀 2. Condicionais — *como uma IA toma decisões*

Todo programa precisa escolher caminhos. A gente faz isso o tempo todo:
- **SE** está chovendo, levo guarda-chuva, **SENÃO** não levo.

```python
if <condição>:
    <faz isso>
elif <outra condição>:
    <faz aquilo>
else:
    <faz o resto>
```
**Regras de ouro:** dois pontos `:` no fim · **4 espaços** de recuo · `elif` e `else` são opcionais.

### 🧩 No Byte: o analisador de sentimento
O Byte conta palavras positivas e negativas e decide o "clima" da mensagem com `if / elif / else`. Isso se chama **análise de sentimento** — uma das tarefas de IA mais usadas por empresas (avaliar reviews, reclamações, comentários em redes sociais).

````python
pontuacao = 2   # ✏️ teste com -3, 0, 5

if pontuacao > 0:
    clima = "positivo 😀"
elif pontuacao < 0:
    clima = "negativo 😟"
else:
    clima = "neutro 😐"

print(f"Pontuação {pontuacao} → {clima}")
````

Agora usando o analisador **de verdade** do Byte:

````python
frases = [
    "amei a aula, foi incrível",
    "odiei esse erro chato",
    "hoje é quinta-feira",
    "não gostei",
]
for frase in frases:
    resultado = byte.analisar_sentimento(byte.tokenizar(frase))
    print(f"{resultado['emoji']}  {frase!r:35} → {resultado['rotulo']}")
````

> 🤔 **Pergunta para a turma:** como o Byte acertou que *"não gostei"* é negativo, se "gostei" é uma palavra positiva? Procure por `"nao" in tokens` no `byte.py`.

### Operadores de comparação

| Operador | Significa | Exemplo |
|---|---|---|
| `==` | igual | `temperatura == 0.7` |
| `!=` | diferente | `intencao != "piada"` |
| `>` `<` | maior / menor | `confianca > 0.5` |
| `>=` `<=` | maior/menor ou igual | `nota >= 7` |

⚠️ `=` **guarda** um valor. `==` **compara**. Trocar um pelo outro é o erro nº 1 de iniciante.

### `and` · `or` · `not` — combinando condições

- **`and`** → as duas precisam ser verdade
- **`or`** → basta uma ser verdade
- **`not`** → inverte

````python
# 🧩 No Byte: a temperatura só é aceita se estiver entre 0 E 1
nova_temperatura = 1.5   # ✏️ teste com 0.3

if nova_temperatura >= 0 and nova_temperatura <= 1:
    print("✅ Temperatura aceita")
else:
    print("❌ Precisa estar entre 0 e 1")
````

## 🧰 3. Funções — *as habilidades da IA*

Uma **função** é um bloco de código com nome, que você chama quando quiser.

> 🍰 **Analogia:** função é uma receita de bolo.
> Nome da receita = nome da função · Ingredientes = **parâmetros** · Modo de preparo = corpo · Bolo pronto = **`return`**

```python
def nome_da_funcao(parametro1, parametro2="valor padrão"):
    resultado = ...
    return resultado
```

````python
def saudar(nome, saudacao="Olá"):
    return f"{saudacao}, {nome}! 👋"

print(saudar("Rafael"))              # usa o valor padrão
print(saudar("Ana", "Bom dia"))      # personalizado
````

### 🧩 No Byte: valor padrão = temperatura!
A função `escolher_resposta` tem `temperatura=TEMPERATURA` como padrão. Temperatura baixa → sempre a mesma resposta. Alta → sorteia. **É exatamente o que a temperatura faz num LLM** (só que lá o sorteio é entre palavras, não entre frases prontas).

````python
respostas = ["Resposta certinha 🤓", "Resposta criativa 🎨", "Resposta maluca 🤪"]

print("Temperatura 0.0:")
for i in range(5):
    print("   ", byte.escolher_resposta(respostas, temperatura=0.0))

print("Temperatura 1.0:")
for i in range(5):
    print("   ", byte.escolher_resposta(respostas, temperatura=1.0))
````

### 🧱 Composição — funções que chamam funções

O "cérebro offline" do Byte é uma função que **não sabe fazer nada sozinha**: ela só chama quem sabe.

```
responder_offline(texto)
 ├── tokenizar(texto)            → quebra a frase em palavras
 ├── detectar_intencao(tokens)   → descobre o que a pessoa quer
 ├── analisar_sentimento(tokens) → descobre o "clima"
 └── escolher_resposta(...)      → sorteia a resposta
```

> 🤖 **Ponte com IA:** isso é a base dos **agentes de IA**. Um agente é um LLM que recebe uma lista de *ferramentas* (funções!) e decide qual chamar. No Byte, o dicionário `COMANDOS` faz esse papel — guarda **funções como valores**:

````python
for nome, info in byte.COMANDOS.items():
    print(f"{nome:<14} → função {info['funcao'].__name__}()")
````

````python
# Chamando uma "ferramenta" pelo nome, igual a um agente faz
ferramenta = byte.COMANDOS["/tabuada"]["funcao"]
ferramenta(["9"], {}, [])
````

⚠️ Função **sem `return`** devolve `None`. Sempre confira!

````python
def somar_errado(a, b):
    a + b            # calcula... e joga fora

def somar_certo(a, b):
    return a + b

print(somar_errado(2, 3))   # None 😱
print(somar_certo(2, 3))    # 5
````

### 🎯 Sua vez (15 min, em dupla) — Classificador de pedidos

Crie a função `classificar_pedido(mensagem)` que devolve **a área responsável** por uma mensagem de cliente:
- tem `"boleto"` **ou** `"pagamento"` → `"💰 financeiro"`
- tem `"erro"` **ou** `"bug"` → `"🛠️ suporte técnico"`
- senão → `"💬 atendimento geral"`

Depois crie `responder_cliente(nome, mensagem)` que **chama** a primeira e devolve uma f-string tipo
`"Olá, Ana! Sua mensagem foi enviada para 💰 financeiro."`

> Isso é **roteamento de intenção** — o primeiro passo de todo chatbot de atendimento.

````python
def classificar_pedido(mensagem):
    mensagem = mensagem.lower()
    if "boleto" in mensagem or "pagamento" in mensagem:
        return "💰 financeiro"
    elif "erro" in mensagem or "bug" in mensagem:
        return "🛠️ suporte técnico"
    else:
        return "💬 atendimento geral"


def responder_cliente(nome, mensagem):
    area = classificar_pedido(mensagem)
    return f"Olá, {nome}! Sua mensagem foi enviada para {area}."


print(responder_cliente("Ana", "Meu boleto não chegou"))
print(responder_cliente("Bruno", "O app deu ERRO ao abrir"))
print(responder_cliente("Carla", "Vocês abrem no sábado?"))
````

## 📚 4. Listas — *como uma IA lê uma frase*

Uma variável guarda **um** valor. Uma **lista** guarda **vários**, em ordem.

> 📦 **Analogia:** se a variável é uma caixa, a lista é uma prateleira de caixas numeradas — e a numeração **começa no zero**.

### 🧩 Tokens: a frase vira lista
Antes de entender qualquer coisa, um LLM quebra seu texto em pedaços chamados **tokens**. O Byte faz uma versão simplificada disso:

````python
tokens = byte.tokenizar("Olá, Byte! Você é uma IA de verdade?")
print(tokens)
print(type(tokens), "com", len(tokens), "itens")
````

> 🤯 O ChatGPT faz a mesma coisa, só que os pedaços são menores que palavras (`"inteligência"` pode virar `["intel", "ig", "ência"]`). E quando falam que um modelo aceita "128 mil tokens", é o **tamanho máximo dessa lista**.

### Índices e fatiamento
| Índice | `[0]` | `[1]` | `[2]` | `[3]` |
|---|---|---|---|---|
| valor | `"ola"` | `"byte"` | `"voce"` | `"e"` |
| negativo | `[-4]` | `[-3]` | `[-2]` | `[-1]` |

````python
print(tokens[0])      # primeiro
print(tokens[-1])     # último
print(tokens[1:3])    # do 1 até o 2 (o 3 não entra!)
print(tokens[:2])     # os 2 primeiros
print(tokens[-3:])    # os 3 últimos
# print(tokens[50])   # ❌ IndexError — descomente para ver
````

### 🧩 Ponte com IA: a *janela de contexto* é um fatiamento
Um LLM não lembra de conversas infinitas. O Byte só manda as **últimas 20 mensagens** para o modelo:
```python
mensagens = [mensagem_sistema] + historico[-MAX_HISTORICO:]
```
É por isso que, em conversas muito longas, o ChatGPT "esquece" o começo 😉

````python
historico_gigante = [f"mensagem {i}" for i in range(1, 101)]   # 100 mensagens
janela = historico_gigante[-5:]
print(janela)
````

### ⚡ List comprehension — filtrar numa linha

> Leia de dentro para fora: *"para cada `t` em `tokens`, **se** `t` estiver nas positivas, pegue `t`"*.

````python
frase = byte.tokenizar("Amei a aula, foi incrível, mas o bug foi chato")

# Forma tradicional
positivas = []
for t in frase:
    if t in byte.PALAVRAS_POSITIVAS:
        positivas.append(t)
print("tradicional:", positivas)

# 🧩 List comprehension — é assim que está no Byte
positivas = [t for t in frase if t in byte.PALAVRAS_POSITIVAS]
negativas = [t for t in frase if t in byte.PALAVRAS_NEGATIVAS]
print("positivas:", positivas)
print("negativas:", negativas)
print("pontuação:", len(positivas) - len(negativas))
````

### 🎯 Sua vez (10 min) — Avaliações de um app

### `while`, `break` e `continue` — *o loop de conversa*

**Todo chatbot do mundo** tem esse esqueleto:

```python
while True:                       # repete para sempre...
    texto = input("Você: ")
    if texto == "":
        continue                  # mensagem vazia? pula e pergunta de novo
    if texto == "sair":
        break                     # ...até alguém dizer "sair"
    resposta = gerar_resposta(texto)
    print(resposta)
```

🧩 Procure a função `conversar()` no `byte.py`: é isso, quase linha por linha.

Como `input()` não funciona bem dentro do notebook, vamos simular o usuário com uma lista:

````python
mensagens_do_usuario = ["oi", "", "me conta uma piada", "", "o que é um token?", "sair", "isso nunca vai aparecer"]

for texto in mensagens_do_usuario:
    if texto == "":
        print("   (mensagem vazia — continue)")
        continue
    if texto == "sair":
        print("   👋 break! Fim da conversa.")
        break
    print(f"Você: {texto}")
````

⚠️ **Cuidado com loop infinito:** `while True` sem `break` nunca para. Se acontecer no terminal: **Ctrl + C**.

> 💡 `for` → quando você sabe quantas vezes vai repetir. `while` → quando repete **até algo acontecer**.

### 🎯 Sua vez (10 min) — Detector de palavrões (guardrail com loop)

Percorra a lista `comentarios`. Para cada um, use `for` + `break` para verificar se **alguma** palavra está em `bloqueadas`. Imprima `✅` ou `🚫` na frente de cada comentário.

````python
bloqueadas = ["idiota", "lixo"]
comentarios = ["adorei o curso", "que lixo de aula", "professor idiota kkk", "muito bom!"]

for comentario in comentarios:
    bloqueado = False
    for palavra in byte.tokenizar(comentario):
        if palavra in bloqueadas:
            bloqueado = True
            break
    simbolo = "🚫" if bloqueado else "✅"
    print(simbolo, comentario)
````

### 🤯 O grande segredo: **uma conversa com o ChatGPT é uma lista de dicionários**

Cada mensagem é um dicionário com `role` (quem falou) e `content` (o que falou). O histórico inteiro é uma lista deles. **Isso é literalmente o formato que a API da OpenAI, Anthropic, Google e cia. recebem.**

````python
historico = [
    {"role": "system",    "content": "Você é um assistente divertido."},
    {"role": "user",      "content": "Oi! Quem é você?"},
    {"role": "assistant", "content": "Sou o Byte 🤖"},
    {"role": "user",      "content": "Me explica o que é uma lista"},
]

for i, msg in enumerate(historico):
    print(f"{i}. [{msg['role']:>9}] {msg['content']}")
````

E quando o Byte usa um LLM, o pedido que vai pela internet é... **um dicionário com uma lista de dicionários dentro**:

````python
import json

pedido = byte.montar_requisicao(historico[1:], temperatura=0.7)
print(json.dumps(pedido, indent=2, ensure_ascii=False))
````

👆 Olha os tipos aí dentro: `str` (model), `float` (temperature), `int` (max_tokens), `list` de `dict` (messages). **Todos os fundamentos de hoje em um único objeto.**

### 🎯 Sua vez (15 min, em dupla) — Relatório da conversa

Crie a função `resumir_conversa(historico)` que recebe uma lista de mensagens e **devolve um dicionário** com:
- `"total"` → número de mensagens
- `"do_usuario"` → quantas têm `role == "user"`
- `"maior_mensagem"` → o texto da mensagem mais longa

````python
def resumir_conversa(historico):
    do_usuario = [m for m in historico if m["role"] == "user"]
    textos = [m["content"] for m in historico]
    maior = ""
    for t in textos:
        if len(t) > len(maior):
            maior = t
    return {"total": len(historico), "do_usuario": len(do_usuario), "maior_mensagem": maior}


print(resumir_conversa(historico))
````
