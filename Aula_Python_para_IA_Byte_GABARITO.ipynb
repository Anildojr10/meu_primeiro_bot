### 🧪 Corretor automático
Alguns exercícios de hoje vêm com **testes**: quando você acerta, aparece ✅; quando erra, aparece ❌ com o valor esperado. Rode a célula abaixo uma vez (você não precisa entender o código dela agora).

````python
def conferir(obtido, esperado, descricao=""):
    if isinstance(esperado, float) and isinstance(obtido, (int, float)):
        acertou = abs(obtido - esperado) < 0.01
    else:
        acertou = obtido == esperado
    if acertou:
        print(f"✅ {descricao}")
    else:
        print(f"❌ {descricao}\n   esperado: {esperado!r}\n   obtido:   {obtido!r}")


conferir(1 + 1, 2, "o corretor está funcionando")
````

## 🔀 2. Condicionais — *como uma IA toma decisões*

Todo programa precisa escolher caminhos. A gente faz isso o tempo todo:
- **SE** está chovendo, levo guarda-chuva, **SENÃO** não levo.

```python
if <condição>:
    <faz isso>
elif <outra condição>:
    <faz aquilo>
else:
    <faz o resto>
```
**Regras de ouro:** dois pontos `:` no fim · **4 espaços** de recuo · `elif` e `else` são opcionais.

### 🧩 No Byte: o analisador de sentimento
O Byte conta palavras positivas e negativas e decide o "clima" da mensagem com `if / elif / else`. Isso se chama **análise de sentimento** — uma das tarefas de IA mais usadas por empresas (avaliar reviews, reclamações, comentários em redes sociais).

````python
pontuacao = 2   # ✏️ teste com -3, 0, 5

if pontuacao > 0:
    clima = "positivo 😀"
elif pontuacao < 0:
    clima = "negativo 😟"
else:
    clima = "neutro 😐"

print(f"Pontuação {pontuacao} → {clima}")
````

Agora usando o analisador **de verdade** do Byte:

````python
frases = [
    "amei a aula, foi incrível",
    "odiei esse erro chato",
    "hoje é quinta-feira",
    "não gostei",
]
for frase in frases:
    resultado = byte.analisar_sentimento(byte.tokenizar(frase))
    print(f"{resultado['emoji']}  {frase!r:35} → {resultado['rotulo']}")
````

### 🔍 Decifrando a f-string
Na célula acima apareceu `{frase!r:35}`. Esses códigos depois dos `:` são só **formatação** (alinhamento, casas decimais). Não precisa decorar, só reconhecer:

| Código | O que faz | Exemplo → resultado |
|---|---|---|
| `!r` | mostra com aspas | `{"oi"!r}` → `'oi'` |
| `:<10` / `:>10` | ocupa 10 espaços, alinhado à esquerda / direita | `{"byte":<10}` → `byte      ` |
| `:.2f` | 2 casas decimais | `{3.14159:.2f}` → `3.14` |
| `:.0%` | vira porcentagem | `{0.85:.0%}` → `85%` |
| `:+d` | mostra o sinal | `{3:+d}` → `+3` |

````python
valor = 0.8571
print(f"[{'byte':<10}] [{'byte':>10}] {valor:.2f} {valor:.0%} {3:+d} {-3:+d} {'oi'!r}")
````

### Operadores de comparação

| Operador | Significa | Exemplo |
|---|---|---|
| `==` | igual | `temperatura == 0.7` |
| `!=` | diferente | `intencao != "piada"` |
| `>` `<` | maior / menor | `confianca > 0.5` |
| `>=` `<=` | maior/menor ou igual | `nota >= 7` |

⚠️ `=` **guarda** um valor. `==` **compara**. Trocar um pelo outro é o erro nº 1 de iniciante.

### `and` · `or` · `not` — combinando condições

- **`and`** → as duas precisam ser verdade
- **`or`** → basta uma ser verdade
- **`not`** → inverte

````python
# 🧩 No Byte: a temperatura só é aceita se estiver entre 0 E 1
nova_temperatura = 1.5   # ✏️ teste com 0.3

if nova_temperatura >= 0 and nova_temperatura <= 1:
    print("✅ Temperatura aceita")
else:
    print("❌ Precisa estar entre 0 e 1")
````

### 🧱 Composição — funções que chamam funções

O "cérebro offline" do Byte é uma função que **não sabe fazer nada sozinha**: ela só chama quem sabe.

```
responder_offline(texto)
 ├── tokenizar(texto)            → quebra a frase em palavras
 ├── detectar_intencao(tokens)   → descobre o que a pessoa quer
 ├── analisar_sentimento(tokens) → descobre o "clima"

```

> 🤖 **Ponte com IA:** isso é a base dos **agentes de IA**. Um agente é um LLM que recebe uma lista de *ferramentas* (funções!) e decide qual chamar. No Byte, o dicionário `COMANDOS` faz esse papel — guarda **funções como valores**:

````python
for nome, info in byte.COMANDOS.items():
    print(f"{nome:<14} → função {info['funcao'].__name__}()")
````

````python
# Chamando uma "ferramenta" pelo nome, igual a um agente faz
ferramenta = byte.COMANDOS["/tabuada"]["funcao"]
ferramenta(["9"], {}, [])
````

### 🎯 Sua vez (15 min, em dupla) — Classificador de pedidos

Crie a função `classificar_pedido(mensagem)` que devolve **a área responsável** por uma mensagem de cliente:
- tem `"boleto"` **ou** `"pagamento"` → `"💰 financeiro"`
- tem `"erro"` **ou** `"bug"` → `"🛠️ suporte técnico"`
- senão → `"💬 atendimento geral"`

Depois crie `responder_cliente(nome, mensagem)` que **chama** a primeira e devolve uma f-string tipo
`"Olá, Ana! Sua mensagem foi enviada para 💰 financeiro."`

> Isso é **roteamento de intenção** — o primeiro passo de todo chatbot de atendimento.

````python
def classificar_pedido(mensagem):
    mensagem = mensagem.lower()
    if "boleto" in mensagem or "pagamento" in mensagem:
        return "💰 financeiro"
    elif "erro" in mensagem or "bug" in mensagem:
        return "🛠️ suporte técnico"
    else:
        return "💬 atendimento geral"


def responder_cliente(nome, mensagem):
    area = classificar_pedido(mensagem)
    return f"Olá, {nome}! Sua mensagem foi enviada para {area}."


# 🧪 Testes

conferir(classificar_pedido("Meu boleto não chegou"), "💰 financeiro", "boleto → financeiro")
conferir(classificar_pedido("Falha no PAGAMENTO"), "💰 financeiro", "PAGAMENTO → financeiro")
conferir(classificar_pedido("O app deu ERRO ao abrir"), "🛠️ suporte técnico", "ERRO → suporte técnico")
conferir(classificar_pedido("Vocês abrem no sábado?"), "💬 atendimento geral", "resto → atendimento geral")
conferir(responder_cliente("Ana", "Meu boleto não chegou"),
         "Olá, Ana! Sua mensagem foi enviada para 💰 financeiro.", "responder_cliente")
````

## 📚 4. Listas — *como uma IA lê uma frase*

Uma variável guarda **um** valor. Uma **lista** guarda **vários**, em ordem.

> 📦 **Analogia:** se a variável é uma caixa, a lista é uma prateleira de caixas numeradas — e a numeração **começa no zero**.

### 🧩 Tokens: a frase vira lista
Antes de entender qualquer coisa, um LLM quebra seu texto em pedaços chamados **tokens**. O Byte faz uma versão simplificada disso:

````python
tokens = byte.tokenizar("Olá, Byte! Você é uma IA de verdade?")
print(tokens)
print(type(tokens), "com", len(tokens), "itens")
````

> 🤯 O ChatGPT faz a mesma coisa, só que os pedaços são menores que palavras (`"inteligência"` pode virar `["intel", "ig", "ência"]`). E quando falam que um modelo aceita "128 mil tokens", é o **tamanho máximo dessa lista**.

### Índices e fatiamento
| Índice | `[0]` | `[1]` | `[2]` | `[3]` |
|---|---|---|---|---|
| valor | `"ola"` | `"byte"` | `"voce"` | `"e"` |
| negativo | `[-4]` | `[-3]` | `[-2]` | `[-1]` |

> 👀 **Veja ao vivo:** abra [platform.openai.com/tokenizer](https://platform.openai.com/tokenizer), cole a mesma frase e compare: quantos tokens o Byte achou e quantos o GPT achou?

### ⚡ List comprehension — filtrar numa linha

> Leia de dentro para fora: *"para cada `t` em `tokens`, **se** `t` estiver nas positivas, pegue `t`"*.

````python
frase = byte.tokenizar("Amei a aula, foi incrível, mas o bug foi chato")

# Forma tradicional
positivas = []
for t in frase:
    if t in byte.PALAVRAS_POSITIVAS:
        positivas.append(t)
print("tradicional:", positivas)

# 🧩 List comprehension — é assim que está no Byte
positivas = [t for t in frase if t in byte.PALAVRAS_POSITIVAS]
negativas = [t for t in frase if t in byte.PALAVRAS_NEGATIVAS]
print("positivas:", positivas)
print("negativas:", negativas)
print("pontuação:", len(positivas) - len(negativas))
````

### 🎯 Sua vez (10 min) — Avaliações de um app

Uma empresa coletou notas (0 a 5) de usuários. Sem loop manual, use `sum()`, `len()`, `max()`, `min()` e list comprehension para calcular:
média, maior, menor, quantas notas ≥ 4 ("promotores") e quantas ≤ 2 ("detratores").

> 📊 Esse tipo de cálculo é o que um time de dados faz antes de treinar um modelo: **entender os dados**.

````python
notas = [5, 4, 2, 5, 3, 1, 4, 5, 5, 2, 4, 3]

media = sum(notas) / len(notas)
maior = max(notas)
menor = min(notas)
promotores = [n for n in notas if n >= 4]
detratores = [n for n in notas if n <= 2]

print("Média:", media, "| Maior:", maior, "| Menor:", menor)
print("Promotores:", promotores, "| Detratores:", detratores)

# 🧪 Testes
conferir(media, 3.58, "média")
conferir(maior, 5, "maior nota")
conferir(menor, 1, "menor nota")
conferir(promotores, [5, 4, 5, 4, 5, 5, 4], "promotores (notas >= 4)")
conferir(detratores, [2, 1, 2], "detratores (notas <= 2)")
````

### 🎯 Sua vez (10 min): detector de palavrões (guardrail com loop)

Complete a função `tem_bloqueada(comentario)`: use `for` + `break` para verificar se **alguma** palavra do comentário está em `bloqueadas`. Ela deve devolver `True` ou `False`.

````python
bloqueadas = ["idiota", "lixo"]
comentarios = ["adorei o curso", "que lixo de aula", "professor idiota kkk", "muito bom!"]


def tem_bloqueada(comentario):
    bloqueado = False
    for palavra in byte.tokenizar(comentario):   # tokenizar já tira maiúsculas e pontuação
        if palavra in bloqueadas:
            bloqueado = True
            break
    return bloqueado


for comentario in comentarios:
    simbolo = "🚫" if tem_bloqueada(comentario) else "✅"
    print(simbolo, comentario)

# 🧪 Testes
conferir(tem_bloqueada("que lixo de aula"), True, "acha 'lixo'")
conferir(tem_bloqueada("Professor IDIOTA!"), True, "maiúsculo e pontuação")
conferir(tem_bloqueada("adorei o curso"), False, "comentário limpo")
````

### 🤯 O grande segredo: **uma conversa com o ChatGPT é uma lista de dicionários**

Cada mensagem é um dicionário com `role` (quem falou) e `content` (o que falou). O histórico inteiro é uma lista deles. **Isso é literalmente o formato que a API da OpenAI, Anthropic, Google e cia. recebem.**

````python
historico = [
    {"role": "system",    "content": "Você é um assistente divertido."},
    {"role": "user",      "content": "Oi! Quem é você?"},
    {"role": "assistant", "content": "Sou o Byte 🤖"},
    {"role": "user",      "content": "Me explica o que é uma lista"},
]

for i, msg in enumerate(historico):
    print(f"{i}. [{msg['role']:>9}] {msg['content']}")
````

### 🧪 Quebre o Byte (5 min)
🔮 **Antes de rodar:** para cada frase abaixo, preveja a intenção e o sentimento que o Byte vai dar. Depois rode e compare.

````python
testes = ["o curso é bom", "não gostei, foi chato", "eu ia sair mais cedo", "minhas senhas são fortes"]

for frase in testes:
    tokens = byte.tokenizar(frase)
    intencao, confianca = byte.detectar_intencao(tokens)
    sentimento = byte.analisar_sentimento(tokens)
    bloqueado = byte.tem_conteudo_bloqueado(tokens)
    print(f"{frase!r:28} → intenção: {intencao:<12} sentimento: {sentimento['rotulo']:<9} bloqueado: {bloqueado}")
````

🤔 **Por que errou?**
- *"o curso é **bom**"* virou **saudação**, porque `"bom"` é gatilho de "bom dia".
- *"não gostei, foi chato"* ficou **neutro**: uma palavra positiva (+1) e uma negativa (−1) se anularam.
- *"eu **ia** sair"* virou assunto de **IA** 😂
- *"minhas **senhas**"* passou pelo guardrail, porque a lista só tem `"senha"` no singular.

Cada erro desses pede **mais uma regra**. E a regra nova cria outro erro... É por isso que regras escritas à mão não escalam.

> ⚖️ **Um minuto de ética:** quem decidiu que `"chato"` é negativo? Fomos nós, numa lista do Python. Quando uma empresa usa IA para classificar comentários de clientes, alguém também escolheu o que conta como "negativo". **Viés começa em decisões assim.**

> 🧠 **Isso é IA?** É a IA "de antigamente": regras e palavras-chave. Funcionou por décadas em chatbots de banco e URA de telefone. Repare que *"qual a capital da França?"* ele não entende — porque **ninguém escreveu essa regra**. Um LLM aprende os padrões sozinho a partir de bilhões de textos. Por isso eles mudaram o jogo.

### 🎯 Sua vez (15 min, em dupla) — Relatório da conversa

Crie a função `resumir_conversa(historico)` que recebe uma lista de mensagens e **devolve um dicionário** com:
- `"total"` → número de mensagens
- `"do_usuario"` → quantas têm `role == "user"`
- `"maior_mensagem"` → o texto da mensagem mais longa

````python
def resumir_conversa(historico):
    do_usuario = [m for m in historico if m["role"] == "user"]
    textos = [m["content"] for m in historico]
    maior = ""
    for t in textos:
        if len(t) > len(maior):
            maior = t
    return {"total": len(historico), "do_usuario": len(do_usuario), "maior_mensagem": maior}


resumo = resumir_conversa(historico)
print(resumo)

# 🧪 Testes
conferir(resumo["total"], 4, "total")
conferir(resumo["do_usuario"], 2, "do_usuario")
conferir(resumo["maior_mensagem"], "Você é um assistente divertido.", "maior_mensagem")
````

### 🔑 Gabarito do desafio 10 (só no arquivo do professor)
**O furo:** em `processar_mensagem()`, a mensagem entra no `historico` **antes** do guardrail rodar. A resposta sai bloqueada, mas a senha fica guardada, e `montar_requisicao()` manda o histórico inteiro para o LLM na mensagem seguinte.

**A correção:** checar o guardrail **antes** do `historico.append(...)`. A célula abaixo aplica a correção no módulo carregado, para o teste dar ✅ sem alterar o `byte.py` do professor. No `byte.py` do aluno, a mudança equivalente vai dentro de `processar_mensagem()`.

````python
def processar_mensagem_corrigida(texto, historico, estado):
    texto = texto.strip()

    if byte.limpar_texto(texto).strip() in ["sair", "tchau", "exit", "quit", "fui"]:
        byte.digitar(f"{byte.NOME_BOT}: Até mais! Foram {len(historico) // 2} trocas de mensagem. Bons códigos! 👋")
        return False

    if texto.startswith("/"):
        byte.executar_comando(texto, estado, historico)
        return True

    # ✅ guardrail ANTES de guardar: o dado sensível nunca entra no histórico
    if byte.tem_conteudo_bloqueado(byte.tokenizar(texto)):
        byte.digitar(f"{byte.NOME_BOT}: 🔒 Opa! Não compartilhe dados sensíveis (senha, cartão, CPF) com uma IA.")
        return True

    historico.append({"role": "user", "content": texto})
    resposta = byte.gerar_resposta(texto, historico, estado)
    historico.append({"role": "assistant", "content": resposta})
    byte.digitar(f"{byte.NOME_BOT}: {resposta}")
    return True


historico_teste = []
estado_teste = {"temperatura": 0.7, "debug": False, "usar_llm": False, "sentimentos": []}
processar_mensagem_corrigida("minha senha é 123456", historico_teste, estado_teste)

vazou = False
for m in historico_teste:
    if "123456" in m["content"]:
        vazou = True
conferir(vazou, False, "a senha NÃO ficou guardada no histórico")
````
